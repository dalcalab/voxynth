```python
import torch
import voxynth
import time
import matplotlib.pyplot as plt
```

```python
displacement = voxynth.transform.random_displacement_field(
    shape=(256, 256, 256), 
    magnitude=torch.tensor([10]).to('cpu'),
    voxsize=1.0,
    device='cpu',
)

displacement = displacement.permute(3, 0, 1, 2).contiguous()
displacement.shape

fig, ax = plt.subplots(1, 3, figsize=(15, 5))

im0 = ax[0].imshow(displacement[0, 128].cpu(), cmap='gray')
plt.colorbar(im0, ax=ax[0])

im1=ax[1].imshow(displacement[1, 128].cpu(), cmap='gray')
plt.colorbar(im1, ax=ax[1])
im2 = ax[2].imshow(displacement[2, 128].cpu(), cmap='gray')
plt.colorbar(im2, ax=ax[2])
```

```python
voxynth.transform.spatial_transform(
    image=torch.randint((2, 256, 256, 256)),
    trf=displacement.unsqueeze(0).repeat(2, 1, 1, 1, 1),
)
```
